```python
from transformers import M2M100ForConditionalGeneration, M2M100Tokenizer
model_name = "facebook/m2m100_418M"
tokenizer = M2M100Tokenizer.from_pretrained(model_name)
model = M2M100ForConditionalGeneration.from_pretrained(model_name)
```

```python
model.save_pretrained("saved_model")
tokenizer.save_pretrained("saved_model")
```

```python
def translate(text, source_lang, target_lang):
    tokenizer.src_lang = source_lang
    encoded = tokenizer(text, return_tensors="pt")
    generated_tokens = model.generate(**encoded, forced_bos_token_id=tokenizer.get_lang_id(target_lang))
    return tokenizer.batch_decode(generated_tokens, skip_special_tokens=True)[0]
```

```python

print("French → Hindi:", translate("Bonjour, comment ça va ?", "fr", "hi"))
print("English → Hindi:", translate("What is your name?", "en", "hi"))
```
